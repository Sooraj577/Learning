```python
# creating a df with non numerical cols
cat_df = data.select_dtypes(exclude="number")
cat_df
```

```python
# checking the missing values
cat_df.isnull().sum()
```

```python
# Finding the count and frequency distribution
for col in cat_df.columns:
    print(col)
    values = list(cat_df[col].value_counts().to_dict().keys())
    count = list(cat_df[col].value_counts().to_dict().values())
    col_df = pd.DataFrame({
        "values": values,
        "count": count,
    })
    col_df['freq(%)'] = col_df['count'] / len(cat_df)
    print(col_df)
    print("\n")
```

```python
# checking the unique values in each column
for i in cat_df.columns:
    print(f"{i}\n{cat_df[i].unique()}\n")
    
```

```python
# columns workclass, occupation & native_country have '?' as value which is invalid and need to be handled appropriately.
print(cat_df['workclass'].value_counts())
cat_df['workclass'].replace("?", np.nan, inplace=True)
print("\nValue counts after replacement\n")
print(cat_df['workclass'].value_counts())
```

```python
print(cat_df['occupation'].value_counts())
cat_df['occupation'].replace('?', np.nan, inplace=True)
print("\nValue counts after replacement\n")
print(cat_df['occupation'].value_counts())
```

```python
print(cat_df['native_country'].value_counts())
cat_df['native_country'].replace('?', np.nan, inplace=True)
print("\nValue count after the replacement.\n")
print(cat_df['native_country'].value_counts())
```
